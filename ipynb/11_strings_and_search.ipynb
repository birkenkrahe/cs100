### Looping over a string

```python
s = "String"
for i in range(4):  # range-based loop (indices)
    print(s[i])
```

```python
for letter in "String": # value-based (no indices)
    print(letter)
```

```python
help(print)
```

```python
for m in s:
    print(m, end=' ')
```

```python
for index in range(len(s)):
    print(s[index], end=' ')
```

### Checking strings in loops

"God is like the sun; you cannot look at it, but without it you cannot look at anything else."

```python
quote = "God is like the sun; you cannot \
look at it, but without it you cannot \
look at anything Else."
```

```python
quote
```

```python
# how many letters `e` are in the quote?
count = 0 # initialize counter
for letter in quote:
    if letter == 'e':
        count = count + 1  # same as `count += 1`
print(count)        
```

```python
# how many letters `e` are in the quote?
count = 0 # initialize counter
for letter in quote:
    if letter == 'e' or 'E': # 'E' is an non-zero ASCII value
        count = count + 1  # same as `count += 1`
print(count)  
```

```python
len(quote)
```

```python
# how many letters `e` are in the quote?
count = 0 # initialize counter
for letter in quote:
    if (letter == 'e') or (letter == 'E'):
        count = count + 1  # same as `count += 1`
print(count) 
```

```python
def how_many_e():
    """ how many letters `e` are in the quote?"""
    count = 0 # initialize counter
    for letter in quote:
        if (letter == 'e') or (letter == 'E'):
            count = count + 1  # same as `count += 1`
    print(count) 

how_many_e()
```

```python
def how_many_e(string):
    """ how many letters `e` are in the quote?"""
    count = 0 # initialize counter
    for letter in string:
        if (letter == 'e') or (letter == 'E'):
            count = count + 1  # same as `count += 1`
    print(count) 

how_many_e("Amen")
```

```python
def how_many_e2(string):
    """ how many letters `e` are in the quote?"""
    count = 0 # initialize counter
    for letter in string:
        if letter in "eE":
            count = count + 1  # same as `count += 1`
    print(count) 

how_many_e2(quote)
```

```python
def how_many_e2(string):
    """ how many letters `e` are in the quote?"""
    count = 0 # initialize counter
    for letter in string:
        if letter.lower() in "e": # turns letter `E` into `e`
            count = count + 1  # same as `count += 1`
    print(count) 

how_many_e2(quote)
```
