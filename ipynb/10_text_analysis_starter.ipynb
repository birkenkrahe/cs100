## Setup

Rebuilds `word_counter`, `print_most_common`, and `subtract` exactly as built in class, so this notebook runs on its own.

Get the text:

```python
import urllib.request
import os

url = "https://www.gutenberg.org/cache/epub/204/pg204.txt"
urllib.request.urlretrieve(url, filename="brown.txt")
print(os.path.getsize("brown.txt"), "bytes")
```

Clean it (dashes, punctuation, lower case):

```python
import unicodedata

filename = "brown.txt"

def split_line(string):
    return string.replace('--', ' ').split()

punc_marks = {}
for line in open(filename):
    for char in line:
        category = unicodedata.category(char)
        if category.startswith('P'):
            punc_marks[char] = 1
punctuation = ''.join(punc_marks)

def clean_word(word):
    return word.strip(punctuation).lower()
```

Build the word-frequency dictionary:

```python
word_counter = {}
for line in open(filename):
    for word in split_line(line):
        word = clean_word(word)
        if word not in word_counter:
            word_counter[word] = 1
        else:
            word_counter[word] += 1

print(len(word_counter))
```

Sorting and printing helpers:

```python
def second_element(term):
    return term[1]

def print_most_common(word_counter, num=5):
    '''print top num most common words in word_counter dictionary'''
    items = sorted(word_counter.items(),
                   key=second_element,
                   reverse=True)
    for word, freq in items[:num]:
        print(f"{word:<10}{freq}")
```

The dictionary-subtraction pattern (needed for Step 2 below):

```python
def subtract(d1, d2):
    res = {}
    for key in d1:
        if key not in d2:
            res[key] = d1[key]
    return res
```
